```python
from google import genai
from dotenv import load_dotenv
import os
```

```python
load_dotenv()

api_key = os.getenv("GEMINI_API_KEY")

client = genai.Client(api_key=api_key)
```

```python
complaint = input("Enter your complaint: ")

print("\nCustomer Input:")
print(complaint)
```

```python
moderation_prompt = f"""
You are a safety moderation system for an e-commerce customer support assistant.

Analyze the following customer input and determine whether it is safe
for the customer support system to process.

Customer input:
{complaint}

Classify the input as either:

SAFE
or
UNSAFE

An input should be considered UNSAFE if it contains:
- Threats of violence
- Instructions for harmful or illegal activity
- Explicit sexual content
- Hate or harassment targeting protected groups
- Requests for dangerous activities

Normal customer complaints, frustration, anger, refund requests,
product complaints, and negative feedback should be considered SAFE.

Return only:

Status: <SAFE or UNSAFE>
Reason: <brief reason>
"""
```

```python
response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=moderation_prompt
)

print(response.text)
```

```python
def moderate_input(complaint):
    
    prompt = f"""
    You are a safety moderation system for an e-commerce customer support assistant.

    Analyze the following customer input:

    {complaint}

    Classify the input as either:

    SAFE
    or
    UNSAFE

    UNSAFE includes:
    - Threats of violence
    - Instructions for harmful or illegal activity
    - Explicit sexual content
    - Hate or harassment targeting protected groups
    - Dangerous activities

    Normal customer complaints, frustration, refund requests,
    product complaints, and negative feedback are SAFE.

    Return only:

    Status: <SAFE or UNSAFE>
    Reason: <brief reason>
    """

    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt
    )

    return response.text
```

```python
complaint = input("Enter your complaint: ")

moderation_result = moderate_input(complaint)

print("\nModeration Result:")
print(moderation_result)
```
