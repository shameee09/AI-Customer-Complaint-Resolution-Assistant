```python
from google import genai
from dotenv import load_dotenv
import os
```

```python
load_dotenv()

api_key = os.getenv("GEMINI_API_KEY")

client = genai.Client(api_key=api_key)
```

```python
complaint = input("Enter your complaint: ")

print("\nCustomer Complaint:")
print(complaint)
```

```python
classification = """
Category: Return/Replacement
Priority: Medium
"""

processed_information = """
Order ID: #12345
Product: black Nike shoe
Issue: Received the wrong size
Customer Request: Exchange it
Date: Last week
"""

resolution = """
Issue: Received the wrong size for order #12345 (black Nike shoe).

Resolution: Process the exchange request for the correct size
pending policy verification.

Reason: The customer received an incorrect size and requested
an exchange, but policy verification is required.
"""

ticket_information = """
Ticket ID: TKT-1001
Status: Pending Review
"""
```

```python
def generate_customer_response(
    complaint,
    classification,
    processed_information,
    resolution,
    ticket_information
):

    prompt = f"""
    You are a professional e-commerce customer support response generator.

    Generate a clear and polite response to the customer.

    Customer Complaint:
    {complaint}

    Complaint Classification:
    {classification}

    Extracted Information:
    {processed_information}

    Resolution:
    {resolution}

    Support Ticket:
    {ticket_information}

    Requirements:

    - Acknowledge the customer's issue.
    - Clearly explain what has been identified.
    - Explain the next step.
    - Mention the support ticket ID.
    - Do not promise a refund, replacement, exchange, or approval
      unless the resolution explicitly confirms it.
    - Do not invent company policies.
    - Keep the response professional and concise.

    Return only the customer-facing response.
    """

    response = client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=prompt
    )

    return response.text
```

```python
customer_response = generate_customer_response(
    complaint,
    classification,
    processed_information,
    resolution,
    ticket_information
)

print("\n CUSTOMER RESPONSE ")
print(customer_response)
```
