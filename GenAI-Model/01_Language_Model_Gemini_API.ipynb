```python
from google import genai
from dotenv import load_dotenv
import os
```

```python
load_dotenv()

api_key = os.getenv("GEMINI_API_KEY")

print("API Key Loaded:", api_key is not None)
```

```python
client = genai.Client(api_key=api_key)

print("Gemini client initialized successfully.")
```

```python
response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents="Explain what Generative AI is in simple words."
)

print(response.text)
```

```python
prompt = """
You are an AI customer support assistant.

A customer has submitted the following complaint:

"My order arrived two days late and the product was damaged. I want a refund."

Explain what the customer's main issue is.
"""

response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=prompt
)

print(response.text)
```

```python
prompt = """
You are an AI Customer Complaint Resolution Assistant.

Analyze the following complaint in 3 short points.

Complaint:
"My order arrived two days late and the product was damaged. I want a refund."
"""

response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents=prompt
)

print(response.text)
```

```python
from google.genai import types
```

```python
response = client.models.generate_content(
    model="gemini-3.5-flash-lite",
    contents="""
    Write a professional response to a customer
    whose order arrived damaged.
    """,
    config=types.GenerateContentConfig(
        temperature=0.3,
        max_output_tokens=150
    )
)

print(response.text)
```
